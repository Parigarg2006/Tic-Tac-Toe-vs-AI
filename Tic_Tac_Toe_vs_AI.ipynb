Tic Tac Toe vs AI

```python
import math

def print_board(board):
    symbols = {1: 'X', -1: 'O', 0: ' '}
    print("\n  1 2 3")
    for i, row in enumerate(board):
        print(f"{i+1} {'|'.join(symbols[c] for c in row)}")
        if i < 2: print("  -+-+-")

def check_winner(board):
    lines = [board[i] for i in range(3)] +             [[board[r][c] for r in range(3)] for c in range(3)] +             [[board[i][i] for i in range(3)], [board[i][2-i] for i in range(3)]]
    for line in lines:
        if sum(line) == 3: return 1
        if sum(line) == -3: return -1
    if all(board[r][c] != 0 for r in range(3) for c in range(3)):
        return 0
    return None

def minimax(board, is_max):
    result = check_winner(board)
    if result is not None: return result
    moves = [(r,c) for r in range(3) for c in range(3) if board[r][c] == 0]
    if is_max:
        best = -math.inf
        for r, c in moves:
            board[r][c] = 1
            best = max(best, minimax(board, False))
            board[r][c] = 0
        return best
    else:
        best = math.inf
        for r, c in moves:
            board[r][c] = -1
            best = min(best, minimax(board, True))
            board[r][c] = 0
        return best

def ai_move(board):
    best_score, best_move = -math.inf, None
    for r in range(3):
        for c in range(3):
            if board[r][c] == 0:
                board[r][c] = 1
                score = minimax(board, False)
                board[r][c] = 0
                if score > best_score:
                    best_score, best_move = score, (r, c)
    return best_move

board = [[0]*3 for _ in range(3)]
print_board(board)
print("\nAI plays as X (1). Human plays as O (-1)")
print("AI makes first move:")
r, c = ai_move(board)
board[r][c] = 1
print_board(board)
winner = check_winner(board)
print(f"\nResult: {winner if winner is not None else 'Game ongoing...'}")
print("\nTo play interactively, wrap in a loop with input() for human moves!")
```
